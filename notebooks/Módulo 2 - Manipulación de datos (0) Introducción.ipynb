### 1.2. Estructuras de datos en Pandas

- **Series**. Array unidimensional etiquetado capaz de almacenar cualquier tipo de dato.
- **DataFrame**. Estructura bidimensional con columnas que pueden ser también de cualquier tipo. Estas columnas son a su vez Series.

### 1.3.  Crear DataFrame

Un DataFrame es la estructura de datos con la que Pandas almacena y manipula datos tabulados (no es la única estructura de datos, pero es la más común). En palabras más sencillas, es como una hoja de cálculo de Excel. Tiene filas y columnas, acepta distintos tipos de datos y permite interactuar entre ellos. La diferencia con una hoja de cálculo es que es extremadamente más rápida y potente, por lo que podrás trabajar con cantidades muy grandes de datos.

Dado que MS Excel solo puede tratar con aproximadamente 1M filas por 16m columnas, la librería pandas se convierte en la ofimática del futuro.

Hay un varias formas de crear un DataFrame con Python desde 0, las más habituales son:

- 1. Diccionario de listas:

````python
data = {"country": ["España", "Australia"],
        "M_population": [47.6, 25.7],
        "M_km2": [0.5, 7.7]}

df = pd.DataFrame(data)
df
````

- 2. Lista de diccionarios:

````python
data = [{"country": "España", "M_population": 47.6, "M_km2": 0.5},
        {"country": "Australia", "M_population": 25.7, "M_km2": 7.7}]

df = pd.DataFrame(data)
df
````

- 3. Lista de listas :

````python
data = [["España", 47.6, 0.5], ["Australia", 25.7, 7.7]]

df = pd.DataFrame(data, columns=["country", "M_population", "M_km2"])
df
````

## 4.  Filtrar DataFrame

No siempre vas a necesitar todo el DataFrame para hacer tus análisis, muchas veces será más cómodo solo tener una parte de él. Para ello podemos hacer subsetting (crear subconjuntos).

En su forma más básica puedes seleccionar las columnas que deseas y filtrar las filas de esta manera:

- ```df['col_name']``` → Selecciona una columna.
- ```df[['col1_name', 'col2_name']]``` → Selecciona 2 o más columnas. Nota que puse una lista dentro de los corchetes.
- ```df[df.col_name == 'value']``` → Filtra los datos.
- ```df[df['col_name'] == 'value']``` → También filtra los datos, pero con distinta notación.
- ```df[df.col_name.isin(list)]``` → Puedes añadir una lista con varios datos a filtrar.
- ```df.query('player_name == value')``` → Otra forma de filtrar datos.

````python
df["Director"]
````

````python
df[["Director", "Title"]]
````

Podemos ver que siempre y cuando nuestra columna no tenga espacios, Pandas permite seleccionar la misma con "." o "[]". Si el nombre de cualquier columna tuviera espacios, es recomendable eliminarlos, sin embargo siempre nos quedaría usar "[]".

````python
df[df.Director == "James Gunn"]
````

````python
df[df["Director"] == "James Gunn"]
````

````python
df[df.Director.isin(["James Gunn", "Billy Ray"])]
````

```df.query()``` es una función que contiene aspectos más avanzados, por lo que la veremos con detenimiento en el módulo 2.

````python
df.query("Director in ('James Gunn', 'Billy Ray')")
````

También puedes usar operadores como "&" o "|" para poner varias condiciones y colocar cada una de ellas entre paréntesis.

````python
df[(df.Director == "James Gunn") & (df.Year == 2014)]
````

````python
df[(df.Director == "James Gunn") | (df.Year == 2014)]
````

Para mejorar el filtrado de strings dentro de una columna se puede utilizar ```df["col_name"].str.lower().str.contains()``` o también ```df["col_name"].str.lower().str.startswith()```. Hay muchas funciones que pueden ser de utilidad, y muchas de ellas las veremos en el módulo 2.

````python
filt = df["Director"].str.lower().str.contains("james")
df[filt]
````

Podemos realizar el contrario del filtro añadiendo ```~``` al principio.

````python
filt = df["Director"].str.lower().str.contains("james")
df[~filt]
````

### 5.1. Método .loc

Ahora que hemos desbloqueado otra funcionalidad de filtrado, podemos usar ```df.loc```:
- ```df.loc[:, ['col_name', 'another_col']]``` → accede a todas las filas y columnas específicas.
- ```df.loc[['row_index', 'another_row']]``` → accede a todas las columnas y filas específicas.
- ```df.loc[['row_index', 'another_row'], ['col_name', 'another_col']]``` → filas y columnas específicas.

Primero consultamos nuestro índice

````python
df.index
````

Fijaros que el primer argumento deben de ser las filas y el siguiente argumento deben de ser las columnas.

````python
df.loc[:, ["Rank", "Director"]]
````

````python
df.loc[["Guardians of the Galaxy", "Sing"]]
````

````python
df.loc[["Guardians of the Galaxy", "Sing"], ["Rank", "Director"]]
````

Podemos eliminar los "[]" y añadir ":" para poder consultar los datos que van desde un rango hasta otro, tanto en columnas como en índices.

````python
df.loc["Guardians of the Galaxy": "Sing", "Rank":"Director"]
````

Utilizando df.loc podemos filtrar de la misma manera que lo hemos hecho anteriormente:

````python
filt = (df.Director == "James Gunn") # Antes
df["Year"][filt]
````

````python
filt = (df.Director == "James Gunn") # Ahora
df.loc[filt, "Year"]
````

Cabe la pena mencionar que no debeís confundir ```df.loc``` con ```df.iloc```. En loc se usan las etiquetas (los nombres asignados tanto a las filas como a las columnas) mientras que en iloc se usan los índices de los elementos (la posición en la fila o la columna, comenzado a contar en 0). Algo que se puede usar para recordar cual se debe usar, al trabajar con índices se debe emplear el que comienza por i.

````python
df.iloc[0:2, 0:2]
````

Para volver al indice original debemos de aplicar:

````python
df.reset_index(inplace = True)
df.head(3)
````

#### 6.2.1 Modificar múltiples filas a la vez

Este cometido lo podemos realizar de múltiples formas:
- Podemos aplicar funciones de texto.
- Podemos aplicar operadores.
- Podemos aplicar o crear nuevas funciones ya existentes y usar apply().
- Podemos usar funciones lambda.
- Podemos usar applymap().
- Podemos usar replace() o map().

Veremos con más con más detalle esta forma de manipulación en el módulo 2.

````python
df1.description = df1.description.str.lower() # Hacer que todas las palabras de la columna description sean minúsculas
df1
````

````python
df1["description_len"] = df1.description.apply(len) # Crear una nueva columna para calcular la longitud del texto
df1
````

````python
def calcular_n_generos(genero):
    genero = genero.split(",")
    genero = len(genero)
    return genero

df1.genre = df1.genre.apply(calcular_n_generos) # Crear una función para contar el número de generos y utilizar .apply
df1
````

````python
df1.runtime_min = df1.runtime_min.apply(lambda x: x/60) # Dividir runtime_min entre 60 para obtener las horas
df1
````

````python
df2 = df1.applymap(lambda x: str(x) + '_X') # Aplicar a todo un dataset una función deseada
df2
````

````python
df1.genre = df1.genre.replace(3, 4) # Remplazamos un elemento tantas veces se repita por otro
df1
````

````python
dict_map = {'James Gunn': 'J. G.',
            'Jorge Rincón': 'J. R.'}

df1.director = df1.director.map(dict_map) # Al igual que replace, reemplazamos los elementos
df1
````

````python
dict_map = {'J. G.': 'James Gunn'}

df1.director = df1.director.map(dict_map) # Tener encuenta la limitación (ver NaN)
df1
````

## 8.  Analizar Series de un DataFrame

Pandas tiene muchas funciones de utilidad para extraer información de las columnas, entre ellas destacan:

- ```.sum()``` → Suma todos los valores de una columna.
- ```.mean()``` → Promedio de los valores de una columna.
- ```.max()``` → Valor máximo de una columna.
- ```.min()``` → Valor mínimo de una columna.

- ```.cummax()``` → En cada fila va poniendo el valor máximo encontrado en orden de una columna.

- ```.unique()``` → Devuelve los valores unicos que existen en una columna.
- ```.nunique()``` → Devuelve el número de valores unicos que existen en una columna.

````python
df["runtime_min"].sum() # Suma del tiempo en minutos
````

## 9. Agrupar datos

Cuando se quiere obtener conclusiones de un dataset, muchas veces es muy interesante poder agrupar la información, esto lo podemos hacer siguiendo las siguientes funciones:

- ```df.groupby('col_name_to_group')['col_name_to_aggregate'].mean()``` → Agrupa los datos de la columna que está entre (paréntesis) y obtiene la media de la columna que está entre [corchetes].
- ```df.groupby('col_name')['col_name'].agg([min, max, sum])``` → Obtiene múltiples estadísticas.
- ```df.groupby(['col_name', 'another_col'])['col_name'].mean()``` → Agrupa varias columnas.
- ```df.groupby(['col_name', 'another_col'])[['col_name', 'another_col']].mean()``` → Obtiene la media de múltiples columnas.

````python
df.groupby("director").mean() # Calcular la media de todas las columnas númericas según los valores únicos de director
````

````python
df.groupby("director")["rating"].mean() # Calcular la media del rating según los valores únicos de director #Siempre teneis la opcion de poner as_index=False como argumento
````

````python
df.groupby("director")["rating"].agg(["min", "max", "mean"]) # Calcular el max, min y media del rating para director
````

````python
df.groupby("director")[["rating", "votes"]].agg({"rating":"min", "votes":"max"}) # Calcular diferentes métricas
````

````python
df.groupby("director", as_index=False)[["rating", "votes"]].agg({"rating":"min", "votes":"max"}) # Sin índices
````

````python
df.groupby("director")["title"].count().loc["Adam McKay"] # Podemos consultar de una "celda" con .loc
````
