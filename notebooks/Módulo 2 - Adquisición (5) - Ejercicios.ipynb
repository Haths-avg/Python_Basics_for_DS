#### 2.1. Haz un request y utiliza los parametros del request para obtener las bebidas que empiecen por la letra a. Después filtra por las bebidas que tengan alcohol ¿Cuantos resultados aparecen?  


```python
# No se puede obtener un listado de bebidas que empiecen con la letra a y que contengan alcohol. La API no lo permite.
# Tenemos que primero que obtener los cócteles que empiecen por la letra a y luego filtrar los que contengan alcohol.

import requests
import json
import pandas as pd

url = "https://www.thecocktaildb.com/api/json/v1/1/search.php?"
querystring = {"f": "a"}

response = requests.get(url, params=querystring)

df = pd.DataFrame(json.loads(response.content))

df_original = df.copy()

df
```

```python
df = df.drinks.apply(pd.Series)
df
```

```python
df.columns
```

```python
# Con el DataFrame anterior, se filtra un listado de columnas donde están los ingredientes para aplicar una
# función que vaya por cada uno de los registros ("celdas") y compruebe si la palabra "juice" este en alguno
# de estos registros. Como resultado se obtiene una matriz booleana, que se convierte en vector con la función
# any().

lista_columnas = ['strIngredient1', 'strIngredient2', 'strIngredient3',
                  'strIngredient4', 'strIngredient5', 'strIngredient6',
                  'strIngredient7', 'strIngredient8', 'strIngredient9',
                  'strIngredient10', 'strIngredient11', 'strIngredient12',
                  'strIngredient13', 'strIngredient14', 'strIngredient15']

def funcion(x):
    if x is not None:
        if "juice" in x.lower():
            return True
        else:
            return False

# resultado = df[["strDrink"] + lista_columnas][df[lista_columnas].applymap(funcion).any(axis=1)]
resultado = df[["strDrink"] + lista_columnas][df[lista_columnas].map(funcion).any(axis=1)]
resultado
```

## 3. Web Scraping

#### 3.1. Crea un scraper (sin usar selenium) para que capture las primeras 10 citas del siguiente sitio web: `https://quotes.toscrape.com/`.

```python
import requests
from bs4 import BeautifulSoup

result = requests.get("https://quotes.toscrape.com/")

soup = BeautifulSoup(result.content, 'html.parser')

frases_html = soup.find_all('span', {'class': 'text'})
frases_texto = [frase.text for frase in frases_html]

for frase in frases_texto:
    print(frase)
```

```python
autores_html = soup.find_all('small', {'class': 'author'})
autores_html
```

```python
autores_texto = [autor.text for autor in autores_html]
autores_texto
```
