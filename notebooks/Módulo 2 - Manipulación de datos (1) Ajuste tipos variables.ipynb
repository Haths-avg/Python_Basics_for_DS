### 1.2. Observar los campos que son enteros:

No es necesario que Pandas use int.64, se puede cambiar:

```python
df.select_dtypes(int).describe()
```

Elegir el int que más interese:

```python
print(np.iinfo(np.int8))
print(np.iinfo(np.int16))
print(np.iinfo(np.int32))
print(np.iinfo(np.int64))
```

Realizar la modificación con `astype()`:

```python
df = df.astype({"id": np.int32,
                "area": np.int16,
                "codigo_delito": np.int16,
                "edad_vict": np.int8})
```

```python
df.info()
```

*****

### 1.3. Observar los campos que son decimales:

Es un falso decimal, realmente se trata de un entero. Pandas ha realizado esto debido a que tiene nulos (`np.nan`).

```python
df.select_dtypes(float).describe()
```

Elegir el float que más nos interese:

```python
print(np.finfo(np.float16))
print(np.finfo(np.float32))
print(np.finfo(np.float64))
```

Una de las opciones es rellenar con un 0 los nulos y convertir la variable a un int:

```python
df["tipo_lugar"] = df["tipo_lugar"].fillna(0).astype(np.int16)
```

*****

### 1.4. Observar los campos que son objectos:

Hay que prestar mucha atención a estos campos:

```python
df.select_dtypes(object).describe()
```

Parece bastante raro que haya 4 categorías en el genero por lo que podemos echar un vistazo:

```python
df["genero_vict"].value_counts()
```

"X" representa "Desconocido" y "H" representa "M" (me lo he inventado, esto se tiene que discutir con la parte de negocio siempre) por lo que aplicamos los cambios:

```python
df["genero_vict"] = df["genero_vict"].fillna("X").str.replace("H", "M")
```

Se puede echar un vistazo a la variable`desc_vict`:

```python
df.desc_vict.value_counts(dropna=False, normalize=True).nsmallest() #nlargest()
```

Una vez hecho esto, se toma ventaja de una funcionalidad que tiene Pandas llamada "categories":

```python
df = df.astype({"desc_vict": 'category',
                "genero_vict": 'category'})
```

## 2. Importante al guardar

Esto es una demonstración de lo que ocurre cuando se guarda un archivo CSV vs Parquet para su posterior análisis:

*Parquet*:

```python
df = pd.read_parquet("https://github.com/JorgeRincon-PY/datasets/raw/main/Crime_LA_2022.parquet")
df.info() #En parquet se guardan los tipos
```

*CSV*:

```python
df = pd.read_csv("https://raw.githubusercontent.com/JorgeRincon-PY/datasets/main/Crime_LA_2022.csv")
df.info() #En CSV no se guardan los tipos
```

Cabe destacar que se pueden convertir los tipos automaticamente, sin embargo, Pandas no lo hace bien:

```python
df.convert_dtypes().dtypes
```
