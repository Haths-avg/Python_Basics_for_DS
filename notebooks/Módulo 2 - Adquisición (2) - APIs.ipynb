## 1. Primera solicitud

Vamos a utilizar esta API para poder entender la librería Requests: https://alexwohlbruck.github.io/cat-facts/docs/

Una solicitud GET es una de las operaciones más comunes cuando se trabaja con APIs. Se trata de obtener una respuesta del servidor con la información que nosotros le hemos pedido:

```python
import requests

response = requests.get('https://cat-fact.herokuapp.com/fact')

print(response)
```

Requests reconoce que los códigos de estado 4XX y 5XX son errores. Por lo tanto, si se muestran estos códigos de estado, el objeto de respuesta de la solicitud se evalúa como False.

```python
if response:
    print('Response OK')
else:
    print('Response Failed')
```

El mensaje “Response Failed” solo aparecerá si se muestra un código de estado 400 o 500. Se puede probar a cambiar la URL por cualquier texto no tenga sentido para ver el fallo de la respuesta y la aparición del estado 404.

Se puede ver el código de estado directamente añadiendo lo siguiente:

```python
print(response.status_code)
```

## 5. Timeout

Timeout en requests es el tiempo máximo que espera la solicitud para recibir una respuesta del servidor antes de cancelar la solicitud y lanzar una excepción de tiempo de espera. Es útil para evitar que una solicitud se bloquee indefinidamente si el servidor no responde o responde muy lentamente.

Por defecto, timeout es None, lo que significa que la solicitud esperará indefinidamente para recibir una respuesta. Al establecer un valor en timeout, la solicitud dejará de esperar después de ese tiempo y lanzará una excepción de tiempo de espera.

```python
result = requests.get("https://cat-fact.herokuapp.com/facts", timeout = 5)
result
```

En este ejemplo, establecemos un tiempo de espera de 5 segundos utilizando el parámetro timeout. Si la solicitud tarda más de 5 segundos en recibir una respuesta del servidor, se lanzará una excepción de tiempo de espera.

Es importante tener en cuenta que establecer un tiempo de espera demasiado corto puede provocar que se cancelen solicitudes que tardan un poco más de lo normal en responder. Por otro lado, un tiempo de espera demasiado largo puede provocar que las solicitudes tardías en responder ralenticen la ejecución de nuestro código.

```python
from requests.exceptions import Timeout

try:
    response = requests.get("https://cat-fact.herokuapp.com/facts", timeout=0.01)
    print(response)

except Timeout:
    print('The request timed out')

else:
    print('The request did not time out')
```

### 6.1. ¿Qué es un decorator?

Una función decoradora ("decorator function" en inglés) es una función que agrega funcionalidad a otra función que recibe como argumento.

Usar una función decoradora es como añadir guacamole a unos nachos. Nos permite agregar funcionalidad a una función existente sin modificarla.

```python
def decorador(f):
    def funcion_nueva():
        print("Funcionalidad extra")
        f()
    return funcion_nueva

@decorador
def funcion_inicial():
    print("Funcionalidad inicial")

funcion_inicial()
```

*****

### 6.2. Reintentar una solicitud con un decorator

Cuando una solicitud falla, es posible que desees que tu aplicación reintenté la misma solicitud. Sin embargo, requests no lo hace por defecto. Para aplicar esta funcionalidad, necesitas implementar código.

Por ejemplo, digamos que deseas que todas las solicitudes a https://cat-fact.herokuapp.com/facts se reintenten tres veces antes de que tu script se de por vencido. Construirías lo siguiente:

```python
import time


def retry(func, retries=3):
    def retry_wrapper(*args, **kwargs):
        attempts = 0
        while attempts < retries:
            try:
                return func(*args, **kwargs)
            except requests.exceptions.RequestException as e:
                print(e)
                time.sleep(2)
                attempts += 1

    return retry_wrapper


@retry
def get_data(url):
    r = requests.get(url)
    return r.text


text = get_data("https://cat-fact.herokuapp.com/facts")
json.loads(text)[1]["text"]
```
