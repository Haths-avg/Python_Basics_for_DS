```python
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.by import By
import time
from bs4 import BeautifulSoup

base_url = "http://books.toscrape.com/"

driver.get(base_url)

items = driver.find_element("xpath", "/html/body/div[1]/div/div/div/form/strong[1]")
n_paginas = int(items.text)/20

lista_links_completa = list()

for i in range(int(n_paginas)):

    soup = BeautifulSoup(driver.page_source, 'html.parser')

    print("Página número ", i+1)

    container = soup.find('ol')
    links = container.find_all('a')
    lista_links = list(set([link.get('href') for link in links]))
    lista_links_casi_completa = ["http://books.toscrape.com/catalogue/" + link.replace("catalogue/", "") for link in lista_links]
    lista_links_completa.append(lista_links_casi_completa)

    if i == 0:
        wait = WebDriverWait(driver, 5).until(EC.presence_of_element_located((By.XPATH, '//*[@id="default"]/div[1]/div/div/div/section/div[2]/div/ul/li[2]/a')))
        boton = driver.find_element(By.XPATH, '//*[@id="default"]/div[1]/div/div/div/section/div[2]/div/ul/li[2]/a')
        boton.click()

    elif i == 49:
        pass

    else:
        wait = WebDriverWait(driver, 5).until(EC.presence_of_element_located((By.XPATH, '//*[@id="default"]/div[1]/div/div/div/section/div[2]/div/ul/li[3]/a')))
        boton = driver.find_element(By.XPATH, '//*[@id="default"]/div[1]/div/div/div/section/div[2]/div/ul/li[3]/a')
        boton.click()
```

```python
lista_links_completa_version2 = [item for sublista in lista_links_completa for item in sublista] #Aplanamos
lista_links_completa_version2 = lista_links_completa_version2[:10] #Cogemos solo 40 items
lista_links_completa_version2
```

```python
import requests

lista_toda_informacion = list()

for item in lista_links_completa_version2:

    print(item)
    result = requests.get(item)

    if result.status_code == requests.codes.ok:

        soup = BeautifulSoup(result.content, 'html.parser')


        nombre = soup.find("h1")
        descripcion = soup.find_all("p")
        atributos_tabla = soup.find_all('tr')

        diccionario = dict()

        diccionario["Nombre"] = nombre.text
        diccionario["Description"] = descripcion[3].text

        for atributo in atributos_tabla:
            diccionario[atributo.select("th")[0].text] = atributo.select("td")[0].text

        lista_toda_informacion.append(diccionario)

    else:
        pass
```
