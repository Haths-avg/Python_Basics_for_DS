## 1. Seleccionar la función / Entender el contexto

Hay cuatro funciones de Pandas que pueden ser utilizadas para fusionar conjuntos de datos. La sintaxis de estas funciones es la siguiente:

******

Siendo dos conjuntos de datos df1y df2:

`df1.append(df2)` :
- Apilado vertical.

`pd.concat([df1, df2])` :
1. Apilado vertical y horizontal.
2. Inner/outer joins en indices.

`df1.join(df2)` :
- inner/outer/left/right joins en indices.

`pd.merge(df1, df2)` :
- Todos los joins en multiples columnas.

*****

#### En la mayoría de los casos siempre se usan `pd.concat()` para apilar y `pd.merge()` para unir.

*****

### 1.1. Apilar

<img src="https://drive.google.com/uc?id=1p8J4xnTYWYYVOuFL-MorquK3y8vf_9Ru" width="60%">

A continuación, se crean dos DataFrames para entender mejor el funcionamiento de la función `.concat()`:

```python
d1 = {"Nombre": ["Jose", "Salva"], "ID": [1, 2]}
d2 = {"Nombre": ["Jorge", "Rubén"], "ID": [3, 4]}

df1 = pd.DataFrame(d1)
df2 = pd.DataFrame(d2)
```

```python
df1
```

```python
df2
```

*****

#### Apilar verticalmente:

```python
pd.concat([df1, df2])
```

Para borrar el indice se puede introducir `ignore_index = True`

```python
pd.concat([df1, df2], ignore_index = True)
```

También se puede crear un nuevo indice para indicar de donde proviene mi conjunto de datos

```python
pd.concat([df1, df2], keys=["A", "B"])
```

*****

#### Apilar horizontalmente:

```python
pd.concat([df1, df2], axis = 1)
```

### 1.2. Unir

<img src="https://drive.google.com/uc?id=1mrnsXv1wjPpBA-cljB2Nc9sX716gqAfK" width="40%">

Hay cuatro tipos de uniones que nos permite realizar la función `.merge()` de Pandas.

- **Inner Join**. Usa la intersección de las claves de ambos conjuntos de datos, similar a un Inner Join de SQL; preserva el orden de las claves de la izquierda.
- **Outer Join**. Usa la unión de las claves de ambos conjuntos de datos, similar a un Full Outer Join de SQL.
- **Left Join**. Utiliza sólo las claves del conjunto de datos izquierdo, similar a una Left Outer Join SQL; conserva el orden de las claves del conjunto izquierdo.
- **Right Join**. Utiliza sólo las claves del marco derecho, de forma similar a una Right Outer Join de SQL; conserva el orden de las claves del conjunto derecho.

El valor por defecto en esta función es el "inner join".

Parametros de la función:

`
pd.merge(
    left,
    right,
    how="inner",
    on=None,
    left_on=None,
    right_on=None,
    left_index=False,
    right_index=False,
    sort=True,
    suffixes=("_x", "_y"),
    copy=True,
    indicator=False,
    validate=None
)
`

Es más sencillo entender los distintos tipos viendo algunos ejemplos:

```python
A = pd.DataFrame({'Nombre': ['Jairo', 'Belen', 'Elena'], 'Pais': ['Spain', 'USA', 'Germany'], 'Puesto': ['CEO', 'CTO', 'CTO']})
A
```

```python
B = pd.DataFrame({'ID': [1, 2, 3], 'Nombre': ['Sheila', 'Rosario', 'Jairo']})
B
```

*****

#### 1.2.1. Inner join

Incluir sólo las observaciones encontradas tanto en A como en B:

```python
pd.merge(A, B, how='inner', on=["Nombre"])
```

*****

#### 1.2.2. Outer join

Incluir las observaciones encontradas en A o B:

```python
pd.merge(A, B, how='outer', on=["Nombre"])
```

*****

####  1.2.3. Left join

Incluir todas las observaciones encontradas en A:

```python
pd.merge(A, B, how='left', on=["Nombre"])
```

*****

#### 1.2.4. Right join

Incluir todas las observaciones encontradas en B:

```python
pd.merge(A, B, how='right', on=["Nombre"])
```

*****

#### 1.2.5. Todo junto

`indicator = True` nos da la funcionalidad de observar que ocurre:

```python
pd.merge(A, B, how="outer", on=["Nombre"], indicator=True)
```
