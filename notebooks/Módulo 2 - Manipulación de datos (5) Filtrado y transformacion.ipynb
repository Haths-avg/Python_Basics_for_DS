## 1. Función `assign()`

La función assign() es útil cuando se desea agregar una o varias columnas calculadas a partir de columnas existentes o agregar columnas que dependen de los datos actuales del DataFrame.

```python
df = pd.DataFrame({'A': [1, 2, 3, 4, 5],
                   'B': [10, 20, 30, 40, 50]})

df_nuevo = df.assign(C=df['A'] + df['B'],  # C es la suma de A y B
                     D=df['A'] * df['B'])  # D es el producto de A y B


df_nuevo
```

Aunque simplemente se puede hacer lo siguiente:

```python
df["C"] = df['A'] + df['B']
df["D"] = df['A'] * df['B']
df
```

### 2.1. Manipulación utilizando regex

La función `.str.extract()`. Para utilizar esta función, debemos definir grupos utilizando paréntesis dentro del patrón. La función extrae las coincidencias y devuelve los grupos como columnas en un dataframe.

- Si se quisiera conseguir el número de producto:

```python
patron = r"^(.{0,3})"

df = df.assign(n_producto = df["Texto"].str.extract(patron))
df
```

- O el precio de cada producto:

```python
patron = r"(\d*,\d\d€/[a-zA-Z]+)"

df = df.assign(precio = df["Texto"].str.extract(patron))
df
```

La función `.replace()`. Esta función sustituye las coincidencias por el string propuesto.

- Si se quiere remplazar una expresión por otra:

```python
patron = r"^(.{0,3})|(\d*,\d\d€/[a-zA-Z]+)"

df = df.assign(producto= df["Texto"].str.replace(patron, "", regex=True))
df
```

## 2.2. Filtrado con Regex

La función `.contains()`. Esta función comprueba un patrón en una columna (Serie) para devolver valores True y False cuando el patrón coincide.

```python
patron = r"100%.*iberico|iberico.*100%"

df[df["Texto"].str.lower().str.contains(patron)]
```

## 5. Mejora de sintaxis

### 5.1. `pipe()`

La función pipe() en Pandas permite aplicar una función (o secuencia de funciones) a un DataFrame o Serie de una manera más limpia y legible. pipe() es especialmente útil cuando se encadenan múltiples operaciones en un solo comando. La idea es que pipe() toma una función y sus argumentos y aplica la función al objeto Pandas (DataFrame o Serie) en cuestión.

```python
df = pd.read_csv("https://raw.githubusercontent.com/JorgeRincon-PY/datasets/main/Titanic.csv")
df
```

- Para todo el dateframe:

```python
def borrar_NAs(df):
    limite = len(df) * 0.6
    df.dropna(axis=1, thresh=limite, inplace=True)
    return df

df.pipe(borrar_NAs)
```

- Para una columna:

```python
def outliers(df, columna):
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1
    filt1 = Q1 - 1.5 * IQR
    filt2 = Q3 + 1.5 * IQR
    return df[df[columna].between(filt1, filt2)].copy()

df.pipe(outliers, "Fare")
```

- Para múltiples columnas:

```python
def borrar(df, columnas):
    for columna in columnas:
        df.drop(columna, axis=1, inplace=True)
    return df

columnas = ["Name", "Embarked"]

df.pipe(borrar, columnas)
```

Usar pipe() puede mejorar la legibilidad del código, especialmente cuando se encadenan múltiples operaciones en un solo comando. También facilita la reutilización de funciones personalizadas en diferentes partes de tu código.

### 5.2. Chaining (encadenamiento)

Se dice que el encadenamiento es más eficiente que los pasos separados, ya que las operaciones pueden ejecutarse consecutivamente sin crear marcos de datos intermedios que aumenten el uso de memoria de Pandas:

```python
df = pd.read_csv("https://raw.githubusercontent.com/JorgeRincon-PY/datasets/main/Titanic.csv")

def borrar_NAs(df):
    limite = len(df) * 0.6
    df.dropna(axis=1, thresh=limite, inplace=True)
    return df

def outliers(df, columna):
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1
    filt1 = Q1 - 1.5 * IQR
    filt2 = Q3 + 1.5 * IQR
    return df[df[columna].between(filt1, filt2)].copy()

def borrar(df, columnas):
    for columna in columnas:
        df.drop(columna, axis=1, inplace=True)
    return df
```

```python
(
    df.pipe(borrar_NAs)
      .pipe(outliers, "Fare")
      .pipe(borrar, ["Name", "Embarked"])
)
```

```python
(
    df.pipe(borrar_NAs)
      .pipe(outliers, "Fare")
      .pipe(borrar, ["Name", "Embarked"])
      .groupby("Sex")["Fare"]
      .agg(["min", "max", "median"])
)
```
