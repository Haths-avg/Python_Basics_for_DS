## 1. Long format vs wide format

Un conjunto de datos puede escribirse en dos formatos diferentes:

- Un formato ancho (wide format) contiene valores que no se repiten en la primera columna.

- Un formato largo (long format) contiene valores que sí se repiten en la primera columna.

## 2. De ancho a largo

La función `.melt()` remodela o transforma un DataFrame existente. Cambia la orientación del DataFrame de un formato ancho a uno largo.

**Ejemplo 1**:

```python
df = pd.DataFrame({'Nueva York': [25], 'Paris': [27], 'Londres': [30]})
df
```

```python
df.melt()
```

**Ejemplo 2**:

```python
df = pd.DataFrame({
    'ciudad': ['Nueva York', 'Londres', 'Paris', 'Berlin', 'Amsterdam'],
    'dia_1': [23, 25, 27, 26, 24],
    'dia_2': [22, 21, 25, 26, 23],
    'dia_3': [26, 25, 24, 27, 23],
    'dia_4': [23, 21, 22, 26, 27],
    'dia_5': [27, 26, 27, 24, 28]
})

df
```

```python
df.melt()
```

No era lo que se buscaba, ¿verdad?. Veamos algunos parámetros de la función melt:

- **id_vars**: Permite usar una/s columna/s como identificadores.

```python
df.melt(id_vars=['ciudad'])
```

- **value_vars**: Permite escoger las columnas con las que queramos operar.

```python
df.melt(id_vars=['ciudad'], value_vars=["dia_1", "dia_2"])
```

- **var_name** y **value_name**: Permiten poner un nombre nuestras nuevas columnas.

```python
df.melt(id_vars=['ciudad'], value_vars=["dia_1", "dia_2"], var_name="dia", value_name="temperatura")
```

## 3. De largo a ancho.

La función `.pivot()` remodela o transforma un DataFrame existente. Cambia la orientación del DataFrame de un formato largo a ancho.

Ejemplo:

```python
df = pd.read_csv("https://github.com/JorgeRincon-PY/datasets/raw/main/stock%20prices.csv")
df
```

```python
len(df)
```

Parámetros:

- **index**: Variable que queremos utilizar para que sus valores únicos sean las filas.
- **columns**: Variable que queremos utilizar para que sus valores únicos sean las columnas.
- **values**: Variable que nos indica los valores del nuevo conjunto de datos.

```python
df.pivot(index='symbol', columns='date', values='close')
```
